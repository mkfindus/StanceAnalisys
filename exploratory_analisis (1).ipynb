# Stance distribution by source

```python
from matplotlib import pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")

stance_mapping = {
    0: "NOT DEFINED",
    1: "Promotional",
    2: "Neutral",
    3: "Discouraging",
    4: "Ambiguous",
    5: "indefinite/ironic",
}

df_stance = df.copy()
df_stance['stance_name'] = df_stance['stance'].apply(lambda x: stance_mapping[x])

def generate_plot(df_param, stance_field = 'stance', groupping_field = None):
    f,ax = plt.subplots()
    if groupping_field:
        sns.countplot(df_param, x='stance',ax = ax, hue = groupping_field)
    else:
        sns.countplot(df_param, x='stance',ax = ax)
    ax.set(xticks=range(6), xticklabels=[stance_mapping[x] for x in range(6)])
    for container in ax.containers:
        ax.bar_label(container)
    plt.xticks(rotation=90)
```

```python
generate_plot(df_stance)
```

```python
generate_plot(df_stance, groupping_field='source')
```

```python
generate_plot(df_stance, groupping_field='gender')
```

```python
generate_plot(df_stance, groupping_field='post type')
```

```python
generate_plot(df_stance, groupping_field='post subtype')
```

```python
generate_plot(df_stance, groupping_field='researcher')
```

```python
stance_mapping = {
    0: "NOT DEFINED",
    1: "Promotional",
    2: "Neutral",
    3: "Discouraging",
    4: "Ambiguous",
    5: "indefinite/ironic",
}
# loading dataframe
df = pd.read_csv('merged_data_pulsar_groundtruth.csv', index_col=None)
df['source'] = df['source'].astype(str)
# to not consider samples with NOT DEFINED or indefinite/ironic class
df = df[~df.stance.isin([0, 5])]
# assigning a classname
df['stance_name'] = df_stance['stance'].apply(lambda x: stance_mapping[x])
df.head()
# selecting only tweets where the 'post subtype' is 'original post' and facebook entries
df_for_classification = df[df['source'].str.contains("Facebook Pages") | (df['source'].str.contains("X") & df['post subtype'].str.contains("original post"))]
```

```python
generate_plot(df_for_classification)
```
